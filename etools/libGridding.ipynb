```python
import os
import subprocess
from pathlib import Path

def saga_adw_minimal(points, field, outdir, bname,
                     template_grid=None, 
                     generate_cv=True,
                     overwrite=False):
    """
    Run SAGA-GIS ADW interpolation with minimal parameters.

    Parameters:
        points (str): Path to input point shapefile or compatible vector.
        field (str): Name of the field to interpolate.
        outdir (str): Directory to write outputs.
        bname (str): Base name for output files.
        template_grid (str or None): Optional reference grid (.sgrd or .tif).
        generate_cv (bool): If True, save cross-validation outputs.
        overwrite (bool): Overwrite existing outputs if True.
    """

    os.makedirs(outdir, exist_ok=True)
    outgrid = Path(outdir) / f"{bname}_adw.sgrd"
    cv_summary = Path(outdir) / f"{bname}_cv_summary.shp"
    cv_residuals = Path(outdir) / f"{bname}_cv_residuals.shp"

    if outgrid.exists() and not overwrite:
        print(f"[SKIP] Output exists: {outgrid}")
        return

    cmd = [
        "saga_cmd", "grid_gridding", "7",
        "-POINTS", str(points),
        "-FIELD", field,
        "-TARGET_OUT_GRID", str(outgrid)
    ]

    if template_grid:
        cmd += ["-TARGET_TEMPLATE", str(template_grid)]

    if generate_cv:
        cmd += [
            "-CV_METHOD", "1",  # leave-one-out
            "-CV_SUMMARY", str(cv_summary),
            "-CV_RESIDUALS", str(cv_residuals)
        ]
    else:
        cmd += ["-CV_METHOD", "0"]

    print("Running SAGA ADW with:")
    print(" ".join(map(str, cmd)))

    subprocess.run(cmd, check=True)
```
