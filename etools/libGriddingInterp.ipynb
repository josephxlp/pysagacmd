```python
import os
import subprocess

def saga_inverse_distance_weighted(
    points,
    field,
    output_grid,
    saga_cmd_path="saga_cmd",
    target_template=None,
    target_definition=0,
    target_user_size=1.0,
    target_user_xmin=0.0,
    target_user_xmax=100.0,
    target_user_ymin=0.0,
    target_user_ymax=100.0,
    target_user_cols=101,
    target_user_rows=101,
    target_user_fits=0,
    search_range=1,
    search_radius=1000.0,
    search_points_all=1,
    search_points_min=1,
    search_points_max=20,
    dw_weighting=1,
    dw_idw_power=2.0,
    dw_bandwidth=1.0,
    cv_method=0,
    cv_samples=10,
    cv_summary=None,
    cv_residuals=None,
    overwrite=False
):
    """
    SAGA GIS: Inverse Distance Weighted Interpolation (grid_gridding 1)

    Parameters:
    - points (str): Path to input point shapefile.
    - field (str): Attribute name (e.g., 'elevation') to interpolate.
    - output_grid (str): Path to output interpolated grid (.sdat).
    - saga_cmd_path (str): Path to `saga_cmd` binary.
    - target_template (str): Path to existing grid to match (optional).
    - target_definition (int): [0] user defined, [1] grid/grid system.
    - target_user_size (float): Output cell size.
    - target_user_xmin/xmax/ymin/ymax (float): Bounding box.
    - target_user_cols/rows (int): Grid dimensions (optional if template given).
    - target_user_fits (int): [0] nodes, [1] cells.
    - search_range (int): [0] local, [1] global.
    - search_radius (float): Max search distance for points.
    - search_points_all (int): [0] nearest N, [1] all in radius.
    - search_points_min/max (int): Limits for used neighbors.
    - dw_weighting (int): [0] none, [1] IDW, [2] exp, [3] Gaussian.
    - dw_idw_power (float): Exponent for IDW.
    - dw_bandwidth (float): For exp/Gaussian.
    - cv_method (int): Cross-validation method [0-3].
    - cv_samples (int): Folds if using k-fold CV.
    - cv_summary (str): Path to write CV summary table (optional).
    - cv_residuals (str): Path to residuals shapefile (optional).
    - overwrite (bool): Skip if output exists unless True.

    Returns:
    - None
    """

    if not overwrite and os.path.exists(output_grid):
        print("Output exists. Skipping. Use overwrite=True to force.")
        return

    cmd = [
        saga_cmd_path, "grid_gridding", "1",
        "-POINTS", points,
        "-FIELD", field,
        "-TARGET_OUT_GRID", output_grid,
        "-TARGET_DEFINITION", str(target_definition),
        "-TARGET_USER_SIZE", str(target_user_size),
        "-TARGET_USER_XMIN", str(target_user_xmin),
        "-TARGET_USER_XMAX", str(target_user_xmax),
        "-TARGET_USER_YMIN", str(target_user_ymin),
        "-TARGET_USER_YMAX", str(target_user_ymax),
        "-TARGET_USER_COLS", str(target_user_cols),
        "-TARGET_USER_ROWS", str(target_user_rows),
        "-TARGET_USER_FITS", str(target_user_fits),
        "-SEARCH_RANGE", str(search_range),
        "-SEARCH_RADIUS", str(search_radius),
        "-SEARCH_POINTS_ALL", str(search_points_all),
        "-SEARCH_POINTS_MIN", str(search_points_min),
        "-SEARCH_POINTS_MAX", str(search_points_max),
        "-DW_WEIGHTING", str(dw_weighting),
        "-DW_IDW_POWER", str(dw_idw_power),
        "-DW_BANDWIDTH", str(dw_bandwidth),
        "-CV_METHOD", str(cv_method),
        "-CV_SAMPLES", str(cv_samples),
    ]

    if target_template:
        cmd += ["-TARGET_TEMPLATE", target_template]
    if cv_summary:
        cmd += ["-CV_SUMMARY", cv_summary]
    if cv_residuals:
        cmd += ["-CV_RESIDUALS", cv_residuals]

    print("Running:", " ".join(cmd))
    subprocess.run(cmd, check=True)
```
